## 2. Selecting Columns by Name

Pandas allows columns to be selected by either by labels or by position.
We can use `[]`, `.loc`, `.at`, `.iloc`, `.iat`.

Inspecting the column names:

````python
X.columns
````

Selecting by name is more explicit. `X.iloc[:, 1]` means "the column currently at position 1", while `X["Length"]` means "the column labeled `Length`".

Selecting the second column by its position:

````python
X.iloc[:, 1]
````

Selecting same column by its label:

````python
X["Length"]
````

If we needed specific rows (No `X[]` equivalent):

````python
X.loc[1:4, "Length"]
````

Multiple columns can be selected by passing a list of labels :

````python
X[["Length", "Diameter", "Height"]]
````

Same result but with `.loc`

````python
X.loc[:, ["Length", "Diameter", "Height"]]
````

The equivalent positional selection is:

````python
X.iloc[:, 1:4]
````

We pick this table as an example for these below :

````python
X.iloc[:, :4]
````

Label-based selection is generally easier to maintain because changing the order of the columns does not change which variables are selected.

We can use `X.at[]` or `X.iat[]` if we needed one specific scalar value : 

````python
X.at[2, "Length"], X.iat[2, 1]
````

````python
X.at[15, "Height"], X.iat[15, 3]
````

> **Note:** `X.at[]` uses **labels**, not integer positions. In this dataset, the row labels happen to be `0, 1, 2, ...`, so `X.at[1, "Length"]` can look similar to positional indexing. However, `1` means **the row whose label is `1`**, not necessarily **the second row**.
>
For example, if the row labels were `[10, 20, 30]`:

 ```python
 X.at[20, "Length"]   # row with label 20
 X.iat[1, 0]          # second row, first column
 ```
 Therefore:

 - `.loc` : label-based, can select multiple cells
 - `.iloc` : position-based, can select multiple cells
 - `.at` : label-based, single cell
 - `.iat` : position-based, single cell


## 3. How Large a Dataset Can Be Loaded This Way?

Pandas loads a DataFrame into memory, so the practical limit depends heavily on the available RAM and the representation of the data.

Important factors include:

- **File size:** how much data must be read from disk or downloaded.
- **Memory representation:** the in-memory DataFrame can require considerably more memory than the original CSV.
- **Processing:** operations such as sorting, joining, copying, or converting data can temporarily require additional memory.
- **Available RAM:** the dataset and intermediate objects must fit comfortably in memory.
- **CPU and storage speed:** larger datasets also take longer to read and process.

A useful experiment is to inspect the memory usage of the current dataset.

````python
df.memory_usage(deep=True)
````

Footprint in MiB:

````python
df.memory_usage(deep=True).sum() / 1024**2, "MiB"
````

There is no single maximum dataset size. A machine with 8 GB of RAM and a machine with 128 GB of RAM have very different practical limits. Furthermore, a DataFrame that occupies 4 GB should not be expected to work comfortably on a machine with only 4 GB of available RAM because Pandas may need additional memory for intermediate operations.

On a server, the same principle applies, but a server may have substantially more RAM and CPU resources. For datasets that exceed the resources of a single machine, distributed or out-of-core tools such as Dask or Spark can be considered.

## 4. Variables With a Very Large Number of Categories

A categorical variable with 100,000 categories could create 100,000 additional columns, It can make the representation very large and can make model training more expensive.

Inspecting the number of categories in the Abalone `Sex` variable :

````python
X["Sex"].nunique()
````

````python
X["Sex"].value_counts()
````

For a variable with many categories, possible approaches include:

- grouping rare categories into an `Other` category;
- using frequency or count encoding;
- using target encoding when appropriate and carefully avoiding data leakage;
- using learned embeddings, especially for neural networks;
- using models that can handle categorical variables directly;
- dropping the variable when it provides little useful information.

### What if every category is unique?

If every row has a unique category, one-hot encoding essentially creates one column per observation. This usually does not provide useful generalizable information.

For example, an identifier such as:

```text
user_000001
user_000002
user_000003
...
```

should normally not be treated as a useful categorical feature merely because it contains strings. It is often an identifier rather than a meaningful predictive variable.

Therefore, a column whose values are all unique should be investigated before being included in a model.

## 5. Alternatives to Pandas

Pandas is useful for tabular data, but it is not the only option.

Some alternatives and related tools include:

| Tool | Typical use |
|---|---|
| NumPy | Numerical arrays and mathematical operations |
| Polars | Fast tabular data processing |
| Dask | Larger-than-memory and parallel DataFrame workloads |
| Apache Spark | Distributed processing of very large datasets |
| PyArrow | Columnar data and efficient data interchange |
| Pillow | Loading and processing images |

For numerical arrays, NumPy can load arrays directly from files:

For images, Pillow provides image-loading functionality:

The appropriate tool depends on the structure and scale of the data. Pandas is particularly convenient for ordinary tabular datasets, while NumPy is more focused on numerical arrays and tools such as Spark are designed for distributed processing.
